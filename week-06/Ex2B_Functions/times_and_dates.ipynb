```python
import datetime
```

# Today

```python
today = datetime.datetime.now()

print("=== Today ===")
print(today.strftime("Today is %A, %B %d, %Y"))
```

# My Birthday

```python
my_birthday = datetime.date(2001, 10, 9)

print("\n=== My Birthday ===")
print(my_birthday.strftime("My birthday is %m/%d/%Y"))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
print("\n=== Next 90 Days ===")
print(ninety_d.strftime("90 days from today is %B %d, %Y"))
```

# Dinner Time

```python
dinner_time = datetime.time(18, 30)
print("\n=== Time for Dinner ===")
print(dinner_time.strftime("Let’s meet for dinner at %I:%M %p"))
```
