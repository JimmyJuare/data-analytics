# Exercise 5.A NumPy Fundamentals

This notebook covers basic NumPy array creation, attributes, and operations.

```python
import numpy as np
```

```python
# a) Create a regular Python list
fruits = ['apple', 'banana', 'cherry', 'date']
print(fruits)
```

```python
# b) Convert list to NumPy array
arr_1 = np.array(fruits)
print(arr_1)
```

```python
# c) Check data type
print(arr_1.dtype)  
# Explanation: <U6 means Unicode string with maximum length of 6 characters
```

```python
# d) Two-dimensional array
arr_2 = np.array([[10, 11, 12], [13, 14, 15]])
print("Array 2:\n", arr_2)
```

```python
# e) Transpose
print("Transposed:\n", arr_2.T)

# Comment: .T returns a transposed view, it does NOT modify the original array
print("\nOriginal arr_2 is unchanged:\n", arr_2)
```

```python
# f) Another 2D array
arr_3 = np.array([[1, 3, 5], [2, 4, 6]])
print(arr_3)
```

```python
# g) Create arr_4 using slicing
arr_4 = np.array([arr_3[0, 2], arr_3[1, 1]])
print(arr_4)
```

```python
# h) Three-dimensional array of ones
arr_5 = np.ones((2, 2, 2), dtype=int)
print(arr_5)
```
