# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code: str) -> str:
    return part_code.split(":", 1)[0]

def get_product_num(part_code: str) -> str:
    return part_code.split(":", 1)[1].split("-", 1)[0]

def get_size(part_code: str) -> str:
    return part_code.split("-", 1)[1]
```

```python
acme = 'ACME:123-L' # supplied by Acme, part #123, size large
di = 'DI:12345-M' # supplied by DI, part #12345, size medium
ace = 'ACE:1-12' # supplied by Ace, part #1, size 12
```

```python
print(get_supplier_code(acme))
print(get_product_num(acme))
print(get_size(acme))
```

```python
print(get_supplier_code(di))
print(get_product_num(di))
print(get_size(di))
```

```python
print(get_supplier_code(ace))
print(get_product_num(ace))
print(get_size(ace))
```
