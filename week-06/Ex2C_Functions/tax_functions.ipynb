## Lab 3. Tax Calculation Functions

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062

def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145

def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:  # 4 or higher
        rate = 0.18 - (0.005 * (withholding_code - 4))
    return gross_pay * rate

print("Person 1:", get_soc_sec_tax(750), get_medicare_tax(750), get_federal_tax(750, 0))
print("Person 2:", get_soc_sec_tax(1550), get_medicare_tax(1550), get_federal_tax(1550, 2))
print("Person 3:", get_soc_sec_tax(1100), get_medicare_tax(1100), get_federal_tax(1100, 5))
```
