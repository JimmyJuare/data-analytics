# Exercise 2.C Creating and calling functions

## Lab 1. Working with custom functions

```python
def favorite_things(name, movie, music):
    print(f"My name is {name}.")
    print(f"My favorite movie is {movie}.")
    print(f"My favorite musician/band is {music}.")

favorite_things("Jimmy", "Inception", "Kendrick Lamar")
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined Year Up United to build a strong foundation in data analysis and launch a meaningful career in tech."))
```

```python
def favorite_place(place, why):
    print(f"One of my favorite places is {place} because {why}.")

favorite_place("Chicago", "of the amazing architecture and vibrant food scene.")
favorite_place("Rocky Mountains", "I love hiking and the peaceful views.")
```
