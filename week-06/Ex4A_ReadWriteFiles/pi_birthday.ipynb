```python
# Read all lines into a list
pi = open('pi_million_digits_bonus.txt', 'r')
pi_lines = pi.readlines()
pi.close()

# Test first few lines
print(pi_lines[0])
print(pi_lines[1])
print(len(pi_lines[0]))
print(len(pi_lines[1]))
```

```python
def pi_bday_check3(bday):
    """Checks if birthday appears in first million digits of pi and shows position."""
    birthday_found = 0
    bday_position = ""
    
    for line in pi_lines:
        if bday in line:
            print(f"Your birthday is in the first million digits of pi!")
            birthday_found += 1
            break
    
    if birthday_found != 1:
        print("Sorry, your birthday was not found in the first million digits of pi")
    else:
        # Build clean pi string without newlines/spaces
        pi_string = ''
        for line in pi_lines:
            pi_string += line.strip()
        
        # Find position (subtract 1 because of the '3.' at the start)
        bday_position = pi_string.index(bday) - 1
        print(f"Your birthday begins at decimal place {bday_position}")

# Test with examples
pi_bday_check3('070281')   # July 2, 1981
pi_bday_check3('100799')   # Oct 7, 1999
pi_bday_check3('081505')   # Aug 15, 2005
pi_bday_check3('110211')   # Should not be found
```
