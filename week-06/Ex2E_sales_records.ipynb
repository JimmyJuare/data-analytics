### Exercise 2.E Using functions for analyzing data

This notebook analyzes a small sales dataset using Python functions. Each function takes the full dataset as its first argument.

```python
sales_data = [
    ('Laptop', 'East', 1200.00),
    ('Phone', 'West', 850.00),
    ('Tablet', 'East', 400.00),
    ('Monitor', 'South', 650.00),
    ('Keyboard', 'West', 75.00),
    ('Laptop', 'South', 1150.00),
    ('Phone', 'East', 900.00),
    ('Tablet', 'West', 425.00),
    ('Monitor', 'East', 680.00),
    ('Keyboard', 'South', 80.00),
    ('Laptop', 'West', 1300.00),
    ('Phone', 'West', 820.00)
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    """Returns the total sales amount for a given region"""
    return sum(amt for _, r, amt in data if r == region)
```

```python
print("East total:", region_total(sales_data, 'East'))
print("West total:", region_total(sales_data, 'West'))
print("South total:", region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    """Returns the number of sales records for a given region"""
    return sum(1 for _, r, _ in data if r == region)
```

```python
print("East count:", region_count(sales_data, 'East'))
print("West count:", region_count(sales_data, 'West'))
print("South count:", region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    """Returns the average sale amount for a region, or 0 if no records"""
    total = region_total(data, region)
    count = region_count(data, region)
    return total / count if count > 0 else 0
```

```python
print("East average:", round(region_average(sales_data, 'East'), 2))
print("West average:", round(region_average(sales_data, 'West'), 2))
print("South average:", round(region_average(sales_data, 'South'), 2))
print("North average (should be 0):", region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    """Returns records where sale amount > threshold"""
    return [record for record in data if record[2] > threshold]

def sales_below(data, threshold):
    """Returns records where sale amount < threshold"""
    return [record for record in data if record[2] < threshold]
```

```python
print("Sales above $500:")
for record in sales_above(sales_data, 500):
    print(record)

print("\nSales below $500:")
for record in sales_below(sales_data, 500):
    print(record)
```

## Lab 5. Top and bottom sale

```python
def top_sale(data):
    """Returns the record with the highest sale amount"""
    return max(data, key=lambda s: s[2])

def bottom_sale(data):
    """Returns the record with the lowest sale amount"""
    return min(data, key=lambda s: s[2])
```

```python
top = top_sale(sales_data)
bottom = bottom_sale(sales_data)

print("Top sale:", top)
print("Bottom sale:", bottom)

# Plain English summary
print(f"\nThe top sale was a {top[0]} in the {top[1]} region for ${top[2]:.2f}.")
print(f"The bottom sale was a {bottom[0]} in the {bottom[1]} region for ${bottom[2]:.2f}.")
```

## Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    """Returns a new list with discounted prices"""
    return list(map(lambda s: (s[0], s[1], round(s[2] * (1 - discount_pct), 2)), data))
```

```python
discounted = apply_discount(sales_data, 0.15)  # 15% discount

print("First 3 records BEFORE discount:")
for record in sales_data[:3]:
    print(record)

print("\nFirst 3 records AFTER 15% discount:")
for record in discounted[:3]:
    print(record)
```

## Lab 7. Unique regions

```python
def unique_regions(data):
    """Returns a sorted list of unique regions"""
    return sorted({r for _, r, _ in data})
```

```python
print("Unique regions:", unique_regions(sales_data))
```

This function is useful when writing longer analysis programs because it helps you dynamically discover what categories (regions, products, departments, etc.) exist in your data without hardcoding them. This pattern is very common in data exploration and reporting.

## Lab 8. Region summary

```python
def region_summary(data, region):
    """Prints a formatted summary report for a region"""
    total = region_total(data, region)
    count = region_count(data, region)
    avg = region_average(data, region)
    
    # Get top sale in this region
    region_sales = [s for s in data if s[1] == region]
    top = max(region_sales, key=lambda s: s[2]) if region_sales else None
    
    print(f"=== {region} Region ===")
    print(f"Records:        {count}")
    print(f"Total Sales:    ${total:,.2f}")
    print(f"Average Sale:   ${avg:,.2f}")
    if top:
        print(f"Top Sale:       {top[0]} — ${top[2]:,.2f}")
    print("-" * 40)
```

```python
region_summary(sales_data, 'East')
region_summary(sales_data, 'West')
region_summary(sales_data, 'South')
```

## Lab 9. Full summary

```python
print("=== FULL SALES SUMMARY ===\n")

for region in unique_regions(sales_data):
    region_summary(sales_data, region)
```
