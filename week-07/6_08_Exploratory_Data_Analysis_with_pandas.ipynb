````python
# 4.1 Games of Chance
import random
# A. Simulate flipping a coin 100 times, then print the percentage of results that were "heads"
def flip_coin(num_flips):
    heads_count = 0
    for _ in range(num_flips):
        if random.choice(['heads', 'tails']) == 'heads':
            heads_count += 1
    percentage_heads = (heads_count / num_flips) * 100
    print(f"Percentage of heads: {percentage_heads:.2f}%")

flip_coin(100)
````

````python
# B. Randomly choose an element until you choose "yellow", incrememting flips each time
#    Print a message each time including how many flips have been executed and what the result was
flips = 0
sides = ["blue", "blue", "blue", "blue", "yellow", "blue"]

def choose_until_yellow():
    global flips
    while True:
        result = random.choice(sides)
        flips += 1
        print(f"Flip {flips}: {result}")
        if result == "yellow":
            break
choose_until_yellow()
````

````python
# C. Simulate a slot machine with the following rules:
#    i. Each turn, the player loses 1 point and three random elements are chosen
#    ii. If the three elements are all the same, the player wins 2 points
#    iii. If all three elements are different, the player wins 5 points
#    Continue taking turns until the player has 0 points or more than 10 points
#    Each turn, print a message describing the outcome of the machine
elements = ["red", "green", "blue"]
player_points = 7

def slot_machine():
    global player_points
    while 0 < player_points <= 10:
        player_points -= 1
        chosen_elements = [random.choice(elements) for _ in range(3)]
        print(f"Chosen elements: {chosen_elements}")
        
        if len(set(chosen_elements)) == 1:  # All the same
            player_points += 2
            print("All the same! You win 2 points.")
        elif len(set(chosen_elements)) == 3:  # All different
            player_points += 5
            print("All different! You win 5 points.")
        else:
            print("No win this turn.")
        
        print(f"Current points: {player_points}\n")
        
slot_machine()
````
