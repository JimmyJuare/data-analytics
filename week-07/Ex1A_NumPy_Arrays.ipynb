# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
# Loop through first dimension (rows)
for x in arrA:
    print(x * 2)
```

```python
# Loop through second dimension (elements)
for x in arrA:
    for y in x:
        print(y * 2)
```

**Difference:**  
The first loop multiplies each row by 2.  
The second loop multiplies each individual element by 2.

```python
# Assign result
for x in arrA:
    loop_result = x * 2
    print(loop_result)
# Comment: loop_result only contains the last row because it gets overwritten each iteration
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)  # Default axis
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# This will cause an error (different shapes)
# stacked_array_2 = np.stack((joined_array_1, joined_array_2))
# Error: All input arrays must have same shape
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# another_2 = np.concatenate((joined_array_1, arrA), axis=1)  # Will error - shapes don't match
```

```python
# hstack, vstack, dstack examples
h = np.hstack((arrA, arrA))      # Horizontal stack (axis=1)
v = np.vstack((arrA, arrA))      # Vertical stack (axis=0)
d = np.dstack((arrA, arrA))      # Depth stack (axis=2)

print("hstack:\n", h)
print("vstack:\n", v)
print("dstack:\n", d)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# Result: 2 arrays, each 2D
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# Result: 3 arrays (last one may be smaller)
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)

y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
