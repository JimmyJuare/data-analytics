# Exercise 1.B NumPy Random

```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)

r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)

r_3 = np.random.rand(2, 1)
print(r_3)

r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)

r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 
                     'Grace', 'Henry', 'Ivy', 'Jack', 'Katie', 'Leo'])

call_on = np.random.choice(students)
print("Called on:", call_on)

breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# Comment: This can repeat names, so not ideal for unique groups.
```

```python
# Weighted selection
call_on_weighted = np.random.choice(students, size=5, 
                                    p=[0.05, 0.1, 0.3, 0.0, 0.15, 0.05, 
                                       0.1, 0.05, 0.05, 0.05, 0.05, 0.05])
print(call_on_weighted)
# p values: Charlie has 30% chance, Diana has 0% chance, etc.
```

```python
# Weighted selection
call_on_weighted = np.random.choice(students, size=5, 
                                    p=[0.05, 0.1, 0.3, 0.0, 0.15, 0.05, 
                                       0.1, 0.05, 0.05, 0.05, 0.05, 0.05])
print(call_on_weighted)
# p values: Charlie has 30% chance, Diana has 0% chance, etc.
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
