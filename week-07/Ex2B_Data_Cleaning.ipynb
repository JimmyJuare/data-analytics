# Exercise 2.B Cleaning up data

```python
# Load the data we created in 2.A
df = pd.read_csv('test_data.csv')
print(df)
print(df.describe())
```

## Filtering a DataFrame

```python
# Age > 18
print(df[df['Age'] > 18])

# Score between 50 and 100
print(df[(df['Score'] > 50) & (df['Score'] < 100)])
```

## Modifying a DataFrame 

```python
df_cleaned = df[df['Score'] >= 0].copy()   # Remove invalid scores
print(df_cleaned)

# Drop City column
df_cleaned = df_cleaned.drop(columns=['City'])
print(df_cleaned)

# Fill missing Age with mean
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].mean())
print(df_cleaned)

# Replace Score == -1 or 0 with mean (if any)
df_cleaned['Score'] = df_cleaned['Score'].replace(-1, df_cleaned['Score'].mean())
print(df_cleaned)

# Drop any remaining NaN
df_cleaned = df_cleaned.dropna()
print(df_cleaned)

# Set ID as index and drop ID column
df_cleaned = df_cleaned.set_index('ID')
print(df_cleaned)
```
