```python
import numpy as np
import matplotlib.pyplot as plt
```

# Exercise 3.A Getting familiar with Matplotlib

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y)
plt.show()
```

```python
# With markers
plt.plot(x, y, marker='o')
plt.show()
```

```python
# With title and labels
plt.plot(x, y, marker='o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
# Dashed red line
plt.plot(x, y, linestyle='--', color='red', marker='o')
plt.title('Styled Line Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Square green markers
plt.plot(x, y, linestyle='-', color='blue', marker='s', 
         markersize=10, markerfacecolor='green')
plt.title('Custom Markers')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# More marker experiments
plt.plot(x, y, linestyle='-.', color='purple', marker='^', markersize=8)
plt.title('Triangle Purple Markers')
plt.show()

plt.plot(x, y, linestyle=':', color='orange', marker='*', markersize=12)
plt.title('Star Orange Markers')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
# Bar Chart
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.bar(products, sales, color='skyblue', width=0.5)
plt.title('Product Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```

```python
# Scatter Plot with variable size and color
x = np.random.rand(50)
y = np.random.rand(50)
sizes = np.random.rand(50) * 100

plt.scatter(x, y, s=sizes, c=sizes, cmap='viridis', alpha=0.7)
plt.title('Scatter Plot with Variable Size & Color')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')
plt.colorbar(label='Size')
plt.show()
```

## Subplots

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Sine wave
x_sine = np.linspace(0, 10, 100)
y_sine = np.sin(x_sine)
axs[0].plot(x_sine, y_sine, color='blue')
axs[0].set_title('Sine Wave')
axs[0].set_xlabel('X')
axs[0].set_ylabel('sin(x)')

# Histogram
data = np.random.randn(1000)
axs[1].hist(data, bins=30, color='green', alpha=0.7)
axs[1].set_title('Normal Distribution')
axs[1].set_xlabel('Value')
axs[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
