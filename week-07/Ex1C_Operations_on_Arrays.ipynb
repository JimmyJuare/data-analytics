```python
import numpy as np
```

```python
## Sample Arrays
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.rand(4, 5)
```

```python
## Permutations
print("Before shuffle:\n", r_4)

np.random.shuffle(r_4)
print("After shuffle:\n", r_4)

new_shuffle = np.random.permutation(r_4)
print("Permutation:\n", new_shuffle)
print("Original after permutation:\n", r_4)
```

```python
## Mathematical Operations
print(r_1 + r_2)   # Works with broadcasting
print(r_1 + r_3)
print(r_2 + r_4)

# Broadcasting examples
print(r_4 + np.array([1]))
print(r_4 + np.array([[1]]))
print(r_4 + np.array([1,1,1,1,1]))
```

Broadcasting allows NumPy to automatically expand smaller arrays to match the shape of larger ones during operations.

```python
## Rounding
print(r_3)
print(np.trunc(r_3))
print(np.fix(r_3))
print(np.floor(r_3))
print(np.ceil(r_3))
print(np.round(r_3, 2))
```

```python
## Cumulative Operations
print(r_4)
print("Total sum:", np.sum(r_4))
print("Sum by rows (axis=0):", np.sum(r_4, axis=0))
print("Sum by columns (axis=1):", np.sum(r_4, axis=1))
print("Cumulative sum:", np.cumsum(r_4))
```

## Grocery Problem


```python
bag1 = np.array([2, 3, 4, 0])   # vegetables, fruits, snacks, beverages
bag2 = np.array([4, 0, 1, 2])

print("Bag 1:", bag1)
print("Bag 2:", bag2)

total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)
print(f"The first bag contains {total_bag1} items, and the second bag contains {total_bag2} items.")

totals = bag1 + bag2
print(f"There are {totals[0]} vegetables, {totals[1]} fruits, {totals[2]} boxes of snacks, and {totals[3]} beverages in the two grocery bags.")
```

## Teamwork


```python
ali = np.array([5, 4, 6, 4, 2])
ben = np.array([0, 5, 5, 3, 4])   # Missed Monday
carla = np.array([3, 6, 0, 3, 4]) # Missed Wednesday

all_knocks = np.array([ali, ben, carla])
print("All knocks:\n", all_knocks)

# Total per person
print("Total doors knocked by each:", np.sum(all_knocks, axis=1))

# Total for the group
print("Group total:", np.sum(all_knocks))

# Rolling total per day
daily_total = np.sum(all_knocks, axis=0)
print("Daily total knocks:", daily_total)

# Estimated votes (1 vote per 2 knocks)
estimated_votes = np.cumsum(daily_total) // 2
print("Cumulative estimated votes:", estimated_votes)
```
